```python
# 1. Import Library
import yaml
import os
import sys

from pathlib import Path
```

```python
# 2. Import Function
sys.path.append(os.path.abspath("../.."))

from src.data_collection import (
    fetch_tickers_data,
    fetch_index_data
)
```

```python
# 3. Load Parameter
with open("../../config/parameter.yaml", "r") as f:
    parameter_config = yaml.safe_load(f)

stocks_parameter = parameter_config["stocks_parameter"]
data_parameter = parameter_config["data_parameter"]

idx_80 = stocks_parameter["idx_80"]
lq_45 = stocks_parameter["lq_45"]
idx_30 = stocks_parameter["idx_30"]
index = stocks_parameter["index"]

interval = data_parameter["interval"]
start_date = data_parameter["start_date"]
end_date = data_parameter["end_date"]

tickers = sorted(set(idx_80 + lq_45 + idx_30))

print(f"Total saham : {len(tickers)}")
print(f"Total index : {len(index)}")
print(f"Interval    : {interval}")
print(f"Periode     : {start_date} to {end_date}")
```

```python
# 4. Retrieve Data
df_all_data, success_tickers = fetch_tickers_data(tickers, start_date, end_date, interval)

df_all_data_index = fetch_index_data(index, start_date, end_date, interval)
```

```python
# 5. Export to CSV
with open("../../config/paths.yaml", "r") as f:
    paths_config = yaml.safe_load(f)

raw_data_path = Path(paths_config["data"]["01_raw"])
raw_data_path.mkdir(parents=True, exist_ok=True)

output_file = raw_data_path / "raw_technicals.csv"
output_file_index = raw_data_path / "raw_technicals_index.csv"

df_all_data.to_csv(output_file, index=False)
df_all_data_index.to_csv(output_file_index, index=False)

print(f"Data saham berhasil disimpan ke : '{output_file}'")
print(f"Data index berhasil disimpan ke : '{output_file_index}'")
print(f"Total saham yang terunduh       : {len(success_tickers)}")
print(f"Total baris data saham          : {len(df_all_data):,}")
print(f"Total baris data index          : {len(df_all_data_index):,}")
```
