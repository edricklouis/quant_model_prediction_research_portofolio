```python
# 1. Import Library
import pandas as pd
import yaml
import os
import sys

from pathlib import Path
```

```python
# 2. Import Function
sys.path.append(os.path.abspath("../.."))

from src.data_preparation import (
    remove_zero_volume_rows
)

from src.features_volume import (
    add_volume_sma,
    add_rolling_vwap
)

from src.features_technical import (
    add_ema_features,
    add_rsi_features,
    add_macd_features,
    add_atr_features,
    add_bollinger_bands_features,
    add_highest,
    add_cmf_features,
    add_mfi_features
)

from src.features_volatility import (
    add_price_level,
    add_current_return,
    add_lagged_returns,
    add_current_volatility,
    add_lagged_volatility,
    add_rolling_skew_kurt,
    add_garman_klass_variance
)

from src.features_normalization import (
    add_volume_spike_ratio,
    add_z_score_volume,
    add_vwap_ratio,
    add_ema_ratio,
    add_macd_hist_norm,
    add_z_score_atr_14,
    add_bb_pct_b,
    add_close_highest_ratio,
    add_z_score_rank_gk_var_21
)

from src.features_statistic import (
    add_rolling_std_21,
    add_z_score_21
)

from src.features_other import (
    add_month,
    add_regime,
    add_rsi_regime,
    add_hidden_demand_supply,
    add_money_flow_magnitude
)
```

```python
# 3. Load Parameter
with open("../../config/parameter.yaml", "r") as f:
    parameter_config = yaml.safe_load(f)

features_parameter = parameter_config["features_parameter"]

round_digits = features_parameter["round_digits"]

print(f"round_digits : {round_digits}")
```

```python
# 4. Load Data
with open("../../config/paths.yaml", "r") as f:
    paths_config = yaml.safe_load(f)

raw_data_path = paths_config["data"]["01_raw"]
csv_file_path = os.path.join(raw_data_path, "raw_technicals.csv")
df_all_data = pd.read_csv(csv_file_path, parse_dates=["date"])
```

```python
# 5. Data Preprocessing
df_all_data = remove_zero_volume_rows(df_all_data)
```

```python
# 6. Features Engineering
df_all_data = add_volume_sma(df_all_data, round_digits)
df_all_data = add_rolling_vwap(df_all_data, round_digits)

df_all_data = add_ema_features(df_all_data, round_digits)
df_all_data = add_rsi_features(df_all_data, round_digits)
df_all_data = add_macd_features(df_all_data, round_digits)
df_all_data = add_atr_features(df_all_data, round_digits)
df_all_data = add_bollinger_bands_features(df_all_data, round_digits)
df_all_data = add_highest(df_all_data)
df_all_data = add_cmf_features(df_all_data, round_digits)
df_all_data = add_mfi_features(df_all_data, round_digits)

df_all_data = add_price_level(df_all_data)
df_all_data = add_current_return(df_all_data)
df_all_data = add_lagged_returns(df_all_data)
df_all_data = add_current_volatility(df_all_data)
df_all_data = add_lagged_volatility(df_all_data)
df_all_data = add_rolling_skew_kurt(df_all_data)
df_all_data = add_garman_klass_variance(df_all_data)

df_all_data = add_volume_spike_ratio(df_all_data)
df_all_data = add_z_score_volume(df_all_data)
df_all_data = add_vwap_ratio(df_all_data)
df_all_data = add_ema_ratio(df_all_data)
df_all_data = add_macd_hist_norm(df_all_data)
df_all_data = add_z_score_atr_14(df_all_data)
df_all_data = add_bb_pct_b(df_all_data)
df_all_data = add_close_highest_ratio(df_all_data)
df_all_data = add_z_score_rank_gk_var_21(df_all_data)

df_all_data = add_rolling_std_21(df_all_data)
df_all_data = add_z_score_21(df_all_data)

df_all_data = add_month(df_all_data)
df_all_data = add_regime(df_all_data)
df_all_data = add_rsi_regime(df_all_data)
df_all_data = add_hidden_demand_supply(df_all_data)
df_all_data = add_money_flow_magnitude(df_all_data)
```

```python
# 7. Export to CSV
with open("../../config/paths.yaml", "r") as f:
    paths_config = yaml.safe_load(f)

features_data_path = Path(paths_config["data"]["02_features"])
features_data_path.mkdir(parents=True, exist_ok=True)

output_file = features_data_path / "features_technicals.csv"
df_all_data.to_csv(output_file, index=False)

print(f"Data berhasil disimpan ke : '{output_file}'")
print(f"Total baris data          : {len(df_all_data):,}")
```
