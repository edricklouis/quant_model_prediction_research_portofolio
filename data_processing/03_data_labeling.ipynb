```python
# 1. Import Library
import pandas as pd
import yaml
import os
import sys

from pathlib import Path
```

```python
# 2. Import Function
sys.path.append(os.path.abspath("../.."))

from src.data_labeling import (
    add_next_close_label,
    add_rank_label
)
```

```python
# 3. Load Parameter
with open("../../config/parameter.yaml", "r") as f:
    parameter_config = yaml.safe_load(f)

features_parameter = parameter_config["features_parameter"]

round_digits = features_parameter["round_digits"]

print(f"round_digits : {round_digits}")
```

```python
# 4. Load Data
with open("../../config/paths.yaml", "r") as f:
    paths_config = yaml.safe_load(f)

features_data_path = paths_config["data"]["02_features"]
csv_file_path = os.path.join(features_data_path, "features_technicals.csv")
df_all_data = pd.read_csv(csv_file_path, parse_dates=["date"])
```

```python
# 5. Labeling
df_all_data = add_next_close_label(df_all_data, round_digits)
df_all_data = add_rank_label(df_all_data)
```

```python
# 6. Export to CSV
with open("../../config/paths.yaml", "r") as f:
    paths_config = yaml.safe_load(f)

labels_data_path = Path(paths_config["data"]["03_labels"])
labels_data_path.mkdir(parents=True, exist_ok=True)

output_file = labels_data_path / "labeled_technicals.csv"
df_all_data.to_csv(output_file, index=False)

print(f"Data saham berhasil disimpan ke : '{output_file}'")
print(f"Total baris data saham          : {len(df_all_data):,}")
```
